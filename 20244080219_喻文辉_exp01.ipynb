实验2：函数、条件、循环与递归

1. 编写score_level(score)函数：90〜100分返回“优秀”，80〜不⾜90分返回
“良好”，60〜不⾜80分返回“及格”，0〜不⾜60分返回“不及格”。对于超出0〜
100范围的数值，抛出ValueError。

```python
def score_level(score):
    if not (0 <= score <=100):
        raise ValueError("分数必须0~100")
    if 90 <= score <=100:
        return "优秀"
    elif 80 <= score <90:
        return "良好"
    elif 60 <= score <80:
        return "及格"
    else:
        return "不及格"

#测试
print(score_level(95))
print(score_level(85))
print(score_level(70))
print(score_level(50))
# print(score_level(101)) #打开会抛出ValueError
```

3. 编写递归版和迭代版阶乘函数，约定0! = 1，输⼊为⾮负整数。⽐较n = 0、1、
5、10时的结果。

```python
#迭代版
def fact_iter(n):
    if n<0:
        raise ValueError("非负整数")
    res = 1
    for i in range(2,n+1):
        res *=i
    return res

#递归版
def fact_recur(n):
    if n<0:
        raise ValueError("非负整数")
    if n ==0:
        return 1
    return n * fact_recur(n-1)

for num in [0,1,5,10]:
    print(f"n={num},迭代={fact_iter(num)},递归={fact_recur(num)}")
```

4. 编写summary(values)函数，接收⾮空数值列表，以元组返回最⼩值、最⼤值和平
均值，并通过元组解包接收结果。

```python
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    mi = min(values)
    ma = max(values)
    avg = sum(values)/len(values)
    return (mi, ma, avg)

res = summary([10,20,30,40])
min_v,max_v,avg_v = res #元组解包
print(min_v,max_v,avg_v)
```

3. 编写parse_student(line)函数：将成绩转换为整数，检查字段数量、姓名是否
为空及成绩是否在0〜100范围内。⽆效记录抛出ValueError。

```python
def parse_student(line):
    line = line.strip()
    if not line:
        raise ValueError("空行")
    parts = line.split(",")
    if len(parts)!=2:
        raise ValueError(f"字段数量错误，得到{len(parts)}个字段")
    name_str, score_str = parts
    name_str = name_str.strip()
    score_str = score_str.strip()
    if len(name_str)==0:
        raise ValueError("姓名为空")
    try:
        score = int(score_str)
    except ValueError:
        raise ValueError(f"成绩无法转为整数:{score_str}")
    if not (0<=score <=100):
        raise ValueError(f"成绩超出范围0‑100:{score}")
    return {"name":name_str, "score":score}
```

4. 按⾏处理⽂件，捕获⽆效记录的异常，记录⾏号与错误原因，并继续处理后续记
录。有效记录保存为字典组成的列表。

```python
valid_records = []
error_log = []
with open(stu_path,"r",encoding="utf‑8") as f:
    lines_all = f.readlines()

for lineno, rawline in enumerate(lines_all, start=1):
    if lineno ==1:
        continue #跳过表头
    try:
        rec = parse_student(rawline)
        valid_records.append(rec)
    except Exception as err:
        error_log.append({"line":lineno, "err_msg":str(err)})

print("有效记录：",valid_records)
print("错误记录：",error_log)
```

5. 将有效记录数、⽆效记录数、平均分和及格率写⼊exp01_data/report.txt。分
别演⽰w与a模式，说明⼆者对已有内容的影响。

```python
valid_num = len(valid_records)
invalid_num = len(error_log)
if valid_num>0:
    avg_all = sum(r["score"] for r in valid_records)/valid_num
    pass_count = sum(1 for r in valid_records if r["score"]>=60)
    pass_rate_report = pass_count / valid_num
else:
    avg_all =0
    pass_rate_report=0

report_path = os.path.join(data_dir,"report.txt")
# w模式：覆盖清空原有文件
with open(report_path,"w",encoding="utf‑8") as fw:
    fw.write(f"有效记录数:{valid_num}\n")
    fw.write(f"无效记录数:{invalid_num}\n")
    fw.write(f"平均分:{avg_all:.2f}\n")
    fw.write(f"及格率:{pass_rate_report:.2%}\n")

# a追加模式，不会覆盖原有，追加一行
with open(report_path,"a",encoding="utf‑8") as fa:
    fa.write("【a模式追加的一行测试文本】\n")
```

`w`打开会直接清空旧文件；`a`是追加，在文件末尾新增内容，不删除原有。

1. 定义Student类，在__init__()中初始化name和score属性，并实现is_passed()
和__str__()⽅法。

```python
class Student:
    def __init__(self,name,score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >=60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score}, passed={self.is_passed()})"
```

3. 定义继承⾃Student的GraduateStudent类，增加research_topic属性，通过
super().__init__()初始化⽗类属性，并重写__str__()。

```python
class GraduateStudent(Student):
    def __init__(self,name,score,research_topic):
        super().__init__(name,score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name},score={self.score},topic={self.research_topic},passed={self.is_passed()})"

g_stu = GraduateStudent("LiHua",88,"AI算法")
print(g_stu)
```

4. 构造包含嵌套列表的对象，例如{"name": "Alice", "scores": [85, 90]}。分
别进⾏直接赋值、copy.copy()和copy.deepcopy()，修改内部列表，⽐较结果。
每次⽐较都从新建的原始对象开始。

```python
original = {"name": "Alice", "scores": [85, 90]}

# 1直接赋值
obj1 = original
obj1["scores"][0]=999
print("直接赋值修改内层列表：orig",original, "obj1",obj1)
original = {"name": "Alice", "scores": [85, 90]} #重置原始

#2 copy.copy浅拷贝
obj2 = copy.copy(original)
obj2["scores"][0]= 777
print("copy.copy浅拷贝：orig",original,"obj2",obj2)
original = {"name": "Alice", "scores": [85, 90]} #重置原始

#3 copy.deepcopy深拷贝
obj3 = copy.deepcopy(original)
obj3["scores"][0]= 555
print("deepcopy深拷贝：orig",original,"obj3",obj3)
```

- 直接赋值：完全同一个对象；内层列表修改两边一起变
- `copy.copy()`浅拷贝：字典外壳新对象，**内层嵌套列表还是共享引用**，修改内层原对象跟着变
- `copy.deepcopy()`：递归全部复制；嵌套列表完全独立，修改副本不会影响原始对象
