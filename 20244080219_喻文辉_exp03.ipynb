实验 1：正则表达式基础匹配

```python
import re

text1 = "cat on mat; later: cats and dogs"
print("====任务1文本：", text1)
res_d = re.findall(r"\d+", text1)
res_s = re.findall(r"\s+", text1)
res_cat_word = re.findall(r"\bcat\b", text1)
print("\\d+匹配结果：", res_d)
print("\\s+匹配结果：", res_s)
print(r"\bcat\b完整单词cat匹配：", res_cat_word)

print("\n====任务2")
text2 = "cat dog bird CAT"
res_choose = re.findall(r"cat|dog", text2)
print("cat|dog：", res_choose)

t = "aaaa"
r1 = re.findall(r"a|aa|aaa", t)
r2 = re.findall(r"aaa|aa|a", t)
print("re.findall(r'a|aa|aaa','aaaa') →", r1)
print("re.findall(r'aaa|aa|a','aaaa') →", r2)

print("\n====任务3 量词测试 ct cat caat caaat")
text3 = "ct cat caat caaat"
print("原始文本：", text3)
print("ca*t  :", re.findall(r"ca*t", text3))
print("ca+t  :", re.findall(r"ca+t", text3))
print("ca?t  :", re.findall(r"ca?t", text3))
print("ca{2}t:", re.findall(r"ca{2}t", text3))
print("ca{1,2}t:", re.findall(r"ca{1,2}t", text3))
```

说明：
导入re模块。测试元字符`\d+`、`\s+`、单词边界`\b`；多选分支；量词`* + ? {m} {m,n}`。
- `\d+`：匹配连续数字
- `\s+`：匹配连续空白字符（空格、换行）
- `\bcat\b`：完整单词边界，只匹配独立单词cat，不匹配cats
正则多选分支**从左向右优先匹配靠前的模式**。
量词：
- `*`：前面字符出现0次或多次
- `+`：前面字符至少出现1次
- `?`：前面字符出现0或1次
- `{n}`：严格n次
- `{n,m}`：n~m次，包含两端


实验 2：分组、回溯引用与替换

```python
import re

#任务1 提取邮箱用户名、域名
mail_text = "联系:alice@example.com;备选:bob@test.org"
print("===任务1 提取邮箱用户名、域名")
pattern_mail = r"(\w+)@([\w\.]+)"
mails = re.findall(pattern_mail, mail_text)
for user, domain in mails:
    print(f"用户名:{user}, 域名:{domain}")

#任务2 后向引用找连续重复单词
print("\n===任务2 后向引用，连续重复单词")
rep_text = "cat cat, dog cat, bird bird"
dup = re.findall(r"(\b\w+\b)\s+\1", rep_text)
print("连续重复的单词：", dup)

#任务3 日期格式转换 年‑月‑日 →月‑日‑年
print("\n===任务3 日期格式转换")
date_raw = """2024-05-01
2024/05/02
2024.05.03"""
print("转换前：")
print(date_raw)
date_pattern = r"(\d{4})([-/.])(\d{2})([-/.])(\d{2})"
date_out = re.sub(date_pattern, r"\3-\5-\1", date_raw)
print("\n转换后(月‑日‑年)：")
print(date_out)

#任务4 电话号码标准化
print("\n===任务4 电话号码格式标准化")
phone_text = "联系我:123-456-7890 或 987.654.3210"
print("原始：", phone_text)
phone_pat = r"(\d{3})[-.](\d{3})[-.](\d{4})"
phone_res = re.sub(phone_pat, r"(\1) \2-\3", phone_text)
print("格式化后：", phone_res)
```

说明：
`()`捕获分组，可以提取子片段；`\数字`为回溯引用，复用前面分组匹配到的内容；
`re.sub(pattern,替换字符串,原文本)`完成文本替换，替换字符串中`\1`代表第1捕获分组。

实验 3 BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body> <h1>数据分析</h1> <p class="course">第六讲:正则表达式</p> <p class="course">第七讲:结构化数据</p> <a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a> </body></html>
"""

soup = BeautifulSoup(html_doc, "html.parser")
#任务1
title_text = soup.title.get_text()
h1_tag = soup.h1
print("===任务1")
print("页面title文本：", title_text)
print("h1标签名称：", h1_tag.name)
print("h1标签文本：", h1_tag.get_text())

#任务2 find_all
print("\n===任务2 find_all查找")
p_course_list = soup.find_all("p", class_="course")
print("class='course'的p段落文本：")
for p in p_course_list:
    print("  ", p.get_text())

a_links = soup.find_all("a")
print("\n所有链接：")
for a in a_links:
    link_text = a.get_text()
    link_href = a["href"]
    print(f" 链接文字:{link_text}, href={link_href}")

#任务3 get_text对比
print("\n===任务3 get_text()对比")
print("整个文档get_text():")
print(soup.get_text())
print("-"*40)
print("单个p段落get_text():")
print(p_course_list[0].get_text())
print("-"*40)
print("单个a链接get_text():")
print(a_links[0].get_text())
```

使用给定HTML字符串，不访问网络。
- BeautifulSoup(html_doc,"html.parser")构建DOM解析树；
- `.name`获取标签名；`.get_text()`提取标签内纯文本，剔除HTML标记；
- `find_all(标签,class_="xxx")`批量查找元素；`a["href"]`读取标签属性。
