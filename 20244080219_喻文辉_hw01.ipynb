问题 2：欧几里得算法求 GCD 最大公约数

```python
def gcd(m,n):
    while n != 0:
        m, n = n, m % n
    return m

#测试三组数字
print(gcd(2018,2019))
print(gcd(1200,300))
print(gcd(5040,60))

#测试负数情况
print(gcd(-12, 8))
```

如果输入负数，取模之后仍然算出正数 gcd；数学上 gcd 一般定义针对非负整数，负数输入得到的是绝对值的最大公约数。

问题 3：近似欧拉数 e

```python
import math

#(1)euler_limit 极限方式近似e
def euler_limit(x):
    return (1 + 1/x)**x

#(2)无穷级数求和前n项，0索引，euler_infinite_sum(0)=0；euler_infinite_sum(1)=1
def euler_infinite_sum(n):
    total = 0
    fact = 1
    for k in range(n):
        if k>0:
            fact = fact * k
        total += 1.0 / fact
    return total

#(3)euler_approx：误差小于epsilon，while循环
def euler_approx(epsilon):
    real_e = math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - real_e) > epsilon:
        if k>0:
            fact *= k
        total += 1.0/fact
        k +=1
    return total

#(4)打印表格函数
def print_euler_sum_table(n):
    for k in range(1, n+1):
        print(euler_infinite_sum(k))

#测试
print("真实e",math.exp(1))
print(euler_infinite_sum(4))
print(euler_approx(0.001))
print_euler_sum_table(5)
```

问题 4：测试整数幂属性

```python
#1.is_power_of_2 判断是否是2的幂，只用除法和取模%
def is_power_of_2(p):
    while p %2 == 0:
        p = p //2
    return p == 1

#2.推广is_power(b,n) 判断n是不是b的幂
def is_power(b, n):
    while n % b == 0:
        n = n // b
    return n ==1

#测试
print(is_power_of_2(8))
print(is_power_of_2(9))
print(is_power(3,81))
print(is_power(3,80))
```
