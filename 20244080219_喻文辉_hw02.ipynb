第一部分：字符串的乐趣

1. `is_palindrome(s)` 判断回文

```python
def is_palindrome(s):
    # 去除空格，转小写
    clean = s.replace(" ", "").lower()
    return clean == clean[::-1]

# 测试
print(is_palindrome("tacocat"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("hello"))
```

2. `is_abecedarian(s)` 判断字母顺序

```python
def is_abecedarian(s):
    clean = s.replace(" ", "").lower()
    for i in range(len(clean)-1):
        if clean[i] > clean[i+1]:
            return False
    return True

#测试
print(is_abecedarian("adder"))
print(is_abecedarian("abcd efgh xyz"))
print(is_abecedarian("dog"))
```

3. `remove_vowels(s)` 删除元音 a e i o u（大小写都删）

```python
def remove_vowels(s):
    vowels = {"a","e","i","o","u","A","E","I","O","U"}
    res = []
    for c in s:
        if c not in vowels:
            res.append(c)
    return "".join(res)

#测试
print(remove_vowels("cat"))    # ct
print(remove_vowels("audio"))  # d
```

第二部分：列表的乐趣

1. `list_reverse(lst)` 返回反转列表；输入不是列表抛错误

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("参数必须是列表")
    return lst[::-1]

#测试
print(list_reverse([1,2,3]))
# list_reverse((1,2,3)) #会触发报错
```

2. `is_sorted(seq)` 判断序列升序，**只遍历一次**

```python
def is_sorted(seq):
    for i in range(len(seq)-1):
        if seq[i] > seq[i+1]:
            return False
    return True

print(is_sorted([1,2,3,4]))
print(is_sorted([1,3,2]))
```

第三部分：元组作为向量、矩阵

1. `vec_scalar_mult(t, s)` 向量乘标量

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    if not isinstance(s, (int, float)):
        raise TypeError("标量必须是整数或浮点数")
    return tuple(x * s for x in t)

print(vec_scalar_mult((1,2,3), 2))
```

2. `vec_inner_product(t1, t2)` 向量内积

```python
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("输入必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个向量维度不一致")
    return sum(a*b for a,b in zip(t1,t2))

print(vec_inner_product((1,2),(3,4))) #1*3+2*4=11
```

3. `check_valid_mx(mx)` 判断是否合法矩阵

```python
def check_valid_mx(mx):
    if not isinstance(mx, tuple):
        return False
    #每一行必须是元组
    for row in mx:
        if not isinstance(row, tuple):
            return False
    if len(mx) == 0:
        return False
    row_len = len(mx[0])
    for row in mx:
        if len(row) != row_len:
            return False
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True

#测试
mx_ok = ((1.0,2),(3,4))
print(check_valid_mx(mx_ok))
```

4. `mx_vec_mult(mx, vec)` 矩阵 × 向量

```python
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("矩阵格式非法")
    if not isinstance(vec, tuple):
        raise TypeError("向量必须是元组")
    n_col = len(mx[0])
    if len(vec) != n_col:
        raise ValueError("矩阵列数与向量维度不一致")
    out = tuple(vec_inner_product(row, vec) for row in mx)
    return out

M = ((1,2),(3,4))
v = (5,6)
print(mx_vec_mult(M, v))
```

第四部分：面向对象 Vector 类

```python
class Vector:
    def __init__(self, dim, elements=None):
        # dim校验
        if not isinstance(dim, int):
            raise TypeError("维度必须为整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        self.dim = dim
        if elements is None:
            self.elements = [0.0]*dim
        else:
            if not isinstance(elements, (list, tuple)):
                raise TypeError("elements必须是列表或元组")
            if len(elements) != dim:
                raise ValueError("元素个数与维度不匹配")
            self.elements = list(elements)

    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("点乘对象必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("向量维度不一致，无法点乘")
        return sum(a*b for a,b in zip(self.elements, other.elements))

    # 比较运算符
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("维度不同不可比较")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector比较")
        if self.dim != other.dim:
            raise ValueError("维度不同不可比较")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector比较")
        if self.dim != other.dim:
            raise ValueError("维度不同不可比较")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector比较")
        if self.dim != other.dim:
            raise ValueError("维度不同不可比较")
        return self.elements > other.elements

    # 乘法 v*x
    def __mul__(self, other):
        #向量 * 向量：逐元素相乘
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("向量维度不一致")
            new_ele = [a*b for a,b in zip(self.elements, other.elements)]
            return Vector(self.dim, new_ele)
        #向量 * 标量
        elif isinstance(other,(int,float)):
            new_ele = [x*other for x in self.elements]
            return Vector(self.dim, new_ele)
        else:
            raise TypeError("不支持该类型相乘")

    # 标量 * v  例如 2*v，调用__rmul__
    def __rmul__(self, other):
        return self * other

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

# =========测试示例========
v1 = Vector(2, [2,3])
v2 = Vector(2, [4,5])
print(v1)
print(v1*v2)    #逐元素相乘
print(2*v1)     #标量左乘
print(v1.dot(v2)) #点积
print(v1 < v2)
```

**Q：维度 dim 和 elements 应该是类属性还是实例属性？为什么？**
A：应该作为**实例属性**。每一个 Vector 向量对象，维度、元素取值都可以不一样；类属性被所有实例共享，如果作为类属性，修改一个向量，会影响全部向量对象。实例属性保证每个向量保存自己独立的数据。
