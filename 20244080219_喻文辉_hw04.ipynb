Python 正则表达式练习

导入正则模块

```python
import re
```

题 1 检查字符串仅包含大小写英文字母、数字 0‑9

```python
def check_alnum(s: str) -> bool:
    # ^从头 $到尾，全部字符都要匹配 [a-zA-Z0-9]
    pattern = r'^[a-zA-Z0-9]+$'
    return bool(re.fullmatch(pattern, s))

# ===== 测试用例 =====
print(check_alnum("Abc123"))   # True
print(check_alnum("Abc_123")) # False 下划线非法
print(check_alnum("12345"))   # True
print(check_alnum(""))        # False 空字符串
```

题 2 匹配 a 后面跟着零个或多个 b

```python
def match_a_zero_b(s: str) -> bool:
    # a 后面 b*：b出现0次及以上
    pat = r'^ab*$'
    return bool(re.fullmatch(pat, s))

# 测试
print(match_a_zero_b("a"))      # True
print(match_a_zero_b("ab"))     # True
print(match_a_zero_b("abb"))    # True
print(match_a_zero_b("acb"))    # False
print(match_a_zero_b("b"))      # False
```

题 3 匹配 a 后面跟着一个或多个 b

```python
def match_a_oneplus_b(s: str) -> bool:
    pat = r'^ab+$'
    return bool(re.fullmatch(pat, s))

# 测试
print(match_a_oneplus_b("a"))     # False
print(match_a_oneplus_b("ab"))    # True
print(match_a_oneplus_b("abbbb")) # True
print(match_a_oneplus_b("ac"))    # False
```

题 4：a 开头，b 结尾，中间任意字符

```python
def match_start_a_end_b(s: str) -> bool:
    pat = r'^a.*b$'
    return bool(re.fullmatch(pat, s))

# 测试
print(match_start_a_end_b("ab"))      # True
print(match_start_a_end_b("a123xb"))  # True
print(match_start_a_end_b("axyz"))    # False，结尾不是b
print(match_start_a_end_b("xabc"))    # False，开头不是a
```

题 5 删除 IP 每段数字前导零；段全 0 保留 1 个 0

```python
def strip_ip_leading_zero(ip_str: str) -> str:
    # 处理每一段数字
    def replace_segment(match):
        num = match.group(0)
        return str(int(num))
    return re.sub(r'\d+', replace_segment, ip_str)

# 测试
print(strip_ip_leading_zero("192.168.001.010")) #192.168.1.10
print(strip_ip_leading_zero("000.010.002.000")) #0.10.2.0
```

题 6：空格和下划线互相交换替换

```python
def swap_space_underscore(s: str) -> str:
    # 先把空格替换成临时符号，下划线替换空格；再临时符号换成下划线
    step1 = re.sub(r' ', '\x01', s)
    step2 = re.sub(r'_', ' ', step1)
    step3 = re.sub(r'\x01', '_', step2)
    return step3

#测试
print(swap_space_underscore("hello world_test")) #hello_world test
```

题 7 yyyy‑mm‑dd → dd‑mm‑yyyy

```python
def convert_date(s: str) -> str:
    pat = r'^(\d{4})-(\d{2})-(\d{2})$'
    res = re.sub(pat, r'\3-\2-\1', s)
    return res

#测试
print(convert_date("2026-11-26")) #26-11-2026
```

题 8 找出所有以 a 或者 e 开头的单词

```python
def find_words_start_a_e(text: str):
    # \b[aeAE]\w* 以a/e开头单词
    return re.findall(r'\b[aeAE]\w*\b', text)

#测试
test_text = "apple egg cat elephant ant book Earth"
print(find_words_start_a_e(test_text))
# ['apple','egg','elephant','ant','Earth']
```

题 9 把字符串所有 Road 缩写成 Rd

```python
def road_to_rd(s: str) -> str:
    return re.sub(r'\bRoad\b', 'Rd', s)

#测试
print(road_to_rd("Main Road, Green Road Roadabc"))
# Main Rd, Green Rd Roadabc
```

题 10 找出长度至少 4 字符的单词

```python
def find_word_ge4(text: str):
    return re.findall(r'\b\w{4,}\b', text)

#测试
s_test = "hi cat hello test a abcd x yz"
print(find_word_ge4(s_test))
# ['hello','test','abcd']
```

题11 删除除字母数字以外所有字符

```python
def keep_alnum_only(s: str) -> str:
    return re.sub(r'[^a-zA-Z0-9]', '', s)

#测试
print(keep_alnum_only("he!ll@o#12$3%")) # hello123
```

题 12 删除括号以及括号内部全部内容

```python
def remove_parenthesis_content(s: str) -> str:
    out = re.sub(r'\(.*?\)', '', s)
    # 把多余空格压缩
    out = re.sub(r'\s+',' ', out).strip()
    return out

#测试
print(remove_parenthesis_content("example (.com)"))
print(remove_parenthesis_content("w3resource"))
print(remove_parenthesis_content("github (.com) stackoverflow (.com)"))
```

网页检索气候数据

导入库 requests + BeautifulSoup

```python
import requests
from bs4 import BeautifulSoup
```

网页访问记录
- 实际使用 URL：https://simple.wikipedia.org/wiki/Buenos_Aires
- 访问日期：2026-11-26
- 说明：页面中气候数据表标题以 "Climate data for" 开头；网络/HTTP 错误由函数直接抛出异常。


任务 1 — 读取 HTML、解析、定位 `<table>`，观察直接子节点结构

```python
url = "https://simple.wikipedia.org/wiki/Buenos_Aires"

headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/130.0.0.0 Safari/537.36"
}

resp = requests.get(url, headers=headers, timeout=15)
resp.raise_for_status()
html = resp.text
soup = BeautifulSoup(html, "html.parser")

all_tables = soup.find_all("table")
climate_table = None

for tbl in all_tables:
    title_text = ""
    cap = tbl.find("caption")
    if cap:
        title_text = cap.get_text()
    else:
        first_th = tbl.find("th")
        if first_th:
            title_text = first_th.get_text()
    clean_title = re.sub(r'\s+', ' ', title_text).strip()
    if clean_title.startswith("Climate data for"):
        climate_table = tbl
        break

print("是否找到气候表：", climate_table is not None)

if climate_table:
    print("=== 表格直接子节点 ===")
    child_count = 0
    for child in climate_table.children:
        if child.name is not None:
            child_count += 1
            print(f"直接子节点 {child_count}：标签名 = {child.name}")
    print(f"直接子节点总数（忽略空白文本）：{child_count}")
```

任务 2 — 编写 `retrieve_climate_table(s)`

```python
def retrieve_climate_table(url: str):
    headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/130.0.0.0 Safari/537.36"
    }
    resp = requests.get(url, headers=headers, timeout=15)
    resp.raise_for_status()  # HTTP 错误直接抛出异常

    soup = BeautifulSoup(resp.text, "html.parser")
    tables = soup.find_all("table")

    for tbl in tables:
        title_text = ""
        cap = tbl.find("caption")
        if cap:
            title_text = cap.get_text()
        else:
            th1 = tbl.find("th")
            if th1:
                title_text = th1.get_text()
        clean_title = re.sub(r'\s+', ' ', title_text).strip()
        if clean_title.startswith("Climate data for"):
            return tbl
    # 页面读取成功，但没有气候数据表
    return None
```

任务 2 测试用例（多个页面，含一个结构有差异的页面）

```python
# 测试1：Buenos Aires，预期返回 table Tag
url_ba = "https://simple.wikipedia.org/wiki/Buenos_Aires"
t1 = retrieve_climate_table(url_ba)
print("Buenos Aires table is None?", t1 is None)

# 测试2：另一个城市（结构有差异），预期也返回 table Tag
url_la = "https://simple.wikipedia.org/wiki/Los_Angeles"
t2 = retrieve_climate_table(url_la)
print("Los Angeles table is None?", t2 is None)

# 测试3：无气候表页面，预期返回 None
url_random = "https://simple.wikipedia.org/wiki/Python_(programming_language)"
t3 = retrieve_climate_table(url_random)
print("Random page table is None?", t3 is None)
```

任务 3 — 编写 `list_climate_table_row_names(s)`

```python
def list_climate_table_row_names(url: str):
    table_tag = retrieve_climate_table(url)
    if table_tag is None:
        return None

    month_set = {
        "Jan", "Feb", "Mar", "Apr", "May", "Jun",
        "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"
    }
    row_names = []
    rows = table_tag.find_all("tr")

    for tr in rows:
        ths = tr.find_all("th")
        if len(ths) == 0:
            continue
        first_th = ths[0]
        txt = first_th.get_text(strip=True)

        if not txt:
            continue
        # 排除标题行
        if txt.startswith("Climate data for"):
            continue
        # 排除月份列标题行
        if txt in month_set:
            continue
        # 排除数据来源行
        if "Source" in txt or "source" in txt:
            continue

        row_names.append(txt)

    return row_names
```

任务 3 测试用例

```python
url_ba = "https://simple.wikipedia.org/wiki/Buenos_Aires"
row_list = list_climate_table_row_names(url_ba)

print("行名列表：")
if row_list is None:
    print("没有找到气候表格")
else:
    for name in row_list:
        print(f"- {name}")
```
